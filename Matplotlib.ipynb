# Sub plot

```python
x= [1,2,3,4,5,6]
y = [5,15,20,25,30,35]
y1 = [i*5 for i in x]
y2 = [i**2 for i in x]
y3 = [i+5 for i in x]
print(x)
print(y)
print(y1)
print(y2)
print(y3)
```

```python
plt.subplot(2,2,1)
plt.plot(x,y,'-')
plt.xlabel("x")
plt.ylabel('y')

plt.subplot(2,2,2)
plt.plot(x,y1,'--')
plt.xlabel('x')
plt.ylabel('y1')

plt.subplot(2,2,3)
plt.plot(x,y2,':')
plt.xlabel('x')
plt.ylabel('y2')

plt.subplot(2,2,4)
plt.plot(x,y3,'-.')
plt.xlabel('x')
plt.ylabel('y3')

plt.show()
plt.subplots_adjust(wspace=200, hspace=0)
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(1, 11)

y = x
y1 = x**2
y2 = x**3
y3 = np.sqrt(x)

plt.subplot(2,2,1)
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel('y')

plt.subplot(2,2,2)
plt.plot(x, y1)
plt.xlabel('x')
plt.ylabel('y1')

plt.subplot(2,2,3)
plt.plot(x, y2)
plt.xlabel('x')
plt.ylabel('y2')

plt.subplot(2,2,4)
plt.plot(x, y3)
plt.xlabel('x')
plt.ylabel('y3')

plt.subplots_adjust(wspace=0.4, hspace=0.6)

plt.show()
```

# Q2 analysis
## 1. arange all ele in ascen order.
## 2. check the len of the data is even or odd. odd is take the middle ele as meadian(Q2) and if it is even pick the middle two values and find the avg of the two values as a median

# Q1 internal mechanism

```python
q1 = np.percentile(stu_score,25)
print(q1)

q2 = np.percentile(stu_score,50)
print(q2)

q3 = np.percentile(stu_score,75)
print(q3)
```

```python
# position = (p/100)*(n-1)
##  p --> percentile
## n --> num of ele
position=(25/100)*(7-1)
print(position) ## 1.5 -->
## Q1 = min_index_posi_val+0.5*(max_index_posi_val - min_index_posi_val)
Q1 = 80+0.5*(85-80)
print(Q1)

## pos = (75/100)*6
Q3 =  95+0.5*(105-95)  ## 100.0
print(Q3)
```

```python
iqr = Q3-Q1
print(iqr)
```

```python
uper_bound_lier = q3+1.5*iqr
print(uper_bound_lier)
```

```python
lower_bound_lier = q1-1.5*iqr
print(lower_bound_lier)
```
