## Histogram: A histogram is a type of chart that shows the frequency distribution of data points across a continuous range of numerical values.

```python
vet = [ 7,18,14,15,26,25,35,29,46,29,61,68,71,74,76,82 ]
bar = [0,10,20,30,40,50,60,70,80]
plt.hist(vet,bar,edgecolor='m')
```

## Pie chart: A pie chart is a circular chart divided into segments (slices) by radii from the center, where each segment illustrates the frequency of a category compared to the whole.

```python
seg_ran = [ 20,33,45,68,59,49 ]
cat = [ 'A','B','C','D','E','F' ]

plt.pie(seg_ran,labels=cat,explode=[0,0,0,0.2,0,0],shadow=True,startangle=45,autopct='%1.1f%%')
plt.legend()
plt.show()
```

```python
seg_ran = [ 20,33,45,68,59,49 ]
cat = [ 'A','B','C','D','E','F' ]

plt.pie(seg_ran,labels=cat,explode=[0,0,0,0.2,0,0],shadow=True,startangle=45,autopct='%1.1f%%',counterclock=False)
plt.legend()
plt.show()
```
