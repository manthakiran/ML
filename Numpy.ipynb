# Aggregate funtions

```python
agg1 = np.random.randint(1,50,15)
print(agg1)
```

```python
print(np.min(agg1))
print(np.max(agg1))
print(np.argmin(agg1))
print(np.argmax(agg1))
print(np.mean(agg1))
print(np.median(agg1))
```

# Joins

```python
arr31 = np.arange(1,10).reshape(3,3)
arr32 = np.arange(1,10).reshape(3,3)
print(arr31)
print(arr32)
```

```python
print(np.concatenate((arr31,arr32),axis=0))
```

```python
print(np.concatenate((arr31,arr32),axis=1))
```
