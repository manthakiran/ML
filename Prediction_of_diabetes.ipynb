```python
def check_corr(size):
    corr = df_diabetes.corr()
    fig,ax = plt.subplots(figsize=(size,size))
    cmap = 'jet'
    ax.matshow(corr,cmap = cmap)
    plt.xticks(range(len(corr.columns)),corr.columns,rotation = 90)
    plt.yticks(range(len(corr.columns)),corr.columns)
    plt.show()
```

```python
from sklearn.impute import SimpleImputer
```

```python
fill_zero = SimpleImputer (missing_values=0,strategy='mean')
x_train = fill_zero.fit_transform(x_train)
x_test = fill_zero.fit_transform(x_test)
x_train[:5]
```

```python
x_test[:5]
```

## Standardize the data

## It is a data processing technique which is used to make all values into a single scale i.e mean is 0 & SD is 1

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
x_test[:5]
```

## Train the model using Naviebayees algorithm

```python
from sklearn.naive_bayes import GaussianNB
```

```python
nb_model = GaussianNB()
nb_model.fit(x_train,y_train.ravel())
```

# Predict the result

```python
from sklearn import metrics
nb_predict_train = nb_model.predict(x_train)
print(f"predicted accuracy of navis bayees model is {metrics.accuracy_score(y_train,nb_predict_train )}")
```

```python
nb_predict_test = nb_model.predict(x_test)
print(f"predicted accuracy of navis bayees model is {metrics.accuracy_score(y_test,nb_predict_test )}")
```

## Confusion Matrix

```python
print("Confusion Matrix")
print( "{}".format(metrics.confusion_matrix(y_test,nb_predict_test)) )
```

## Classification Report

```python
print("Classification Matrix")
print( "{}".format(metrics.classification_report(y_test,nb_predict_test)) )
```

## Train the model using Random Forest 

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(x_train,y_train.ravel())
```

```python
from sklearn import metrics

rf_pred_train = rf_model.predict(x_train)
print(f"predicted accuracy of random forest model is {metrics.accuracy_score(y_train,rf_pred_train )}")
```

```python
rf_pred_test = rf_model.predict(x_test)
print(f"predicted accuracy of random forest model is {metrics.accuracy_score(y_test,rf_pred_test )}")
```

```python
print("Confusion Matrix")
print( "{}".format(metrics.confusion_matrix(y_test,rf_pred_test)) )
```
