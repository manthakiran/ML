```python
fill_zero = SimpleImputer (missing_values=0,strategy='mean')
x_train = fill_zero.fit_transform(x_train)
x_test = fill_zero.fit_transform(x_test)
x_train[:5]
```

```python
(x_train == 0).sum()
```

# Standardizing

```python
# It is a data processing technique which is used to make all values into a single scale i.e mean is 0 & SD is 1

### z-score standardization = (data_point1 - Column_mean) + (data_point1 - Column_mean) +....+ (data_point1 - Column_mean)          
###                           -------------------------------------------------------------------------------------------
###                                                           Standard_deviation
### Standard_deviation = Square root of Variance

### Variance = ((data_point1 - Column_mean)**2) + ((data_point2 - Column_mean)**2) +...+ ((data_point_n - Column_mean)**2)
###            -----------------------------------------------------------------------------------------------------------
###                                               len(column)
```

```python
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
x_train[:5]
```

# Training the 3 different models

```python
# Traning the Naive Bayes model

nb_model = GaussianNB()
nb_model.fit(x_train,y_train.ravel())
```

```python
# Traning the Random Forest model

rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(x_train,y_train.ravel())
```

```python
# Traning the KNN model

knn_model = KNeighborsClassifier()
knn_model.fit(x_train,y_train.ravel())
```

# Testing on Train Data 

```python
# Testing the Naive Bayes model using train data
nb_predict_train = nb_model.predict(x_train)

# checking Naive Bayes Model Accuracy on train data
print(f"predicted accuracy of navis bayees model is {metrics.accuracy_score(y_train,nb_predict_train )*100}")
```

```python
# Testing the Random Forest model using train data
rf_pred_train = rf_model.predict(x_train)

# checking Random Forest Model Accuracy on train data
print(f"predicted accuracy of random forest model is {metrics.accuracy_score(y_train,rf_pred_train )*100}")
```

```python
# Testing the KNN model using train data
knn_pred_train = knn_model.predict(x_train)

# checking KNN Model Accuracy on train data
print(f"predicted accuracy of KNN model is {metrics.accuracy_score(y_train,knn_pred_train )*100}")
```

# Testing on Test Data

```python
# Testing the Naive Bayes model using test data
nb_predict_test = nb_model.predict(x_test)

# checking Naive Bayes Model Accuracy on test data
print(f"predicted accuracy of navis bayees model is {metrics.accuracy_score(y_test,nb_predict_test )*100}")
```

```python
# Testing the Random Forest model using test data
rf_pred_test = rf_model.predict(x_test)

# checking Random Forest model Accuracy on test data
print(f"predicted accuracy of random forest model is {metrics.accuracy_score(y_test,rf_pred_test )*100}")
```

```python
# Testing the KNN model using test data
knn_pred_test = knn_model.predict(x_test)

# checking KNN Model Accuracy on test data
print(f"predicted accuracy of KNN model is {metrics.accuracy_score(y_test,knn_pred_test )*100}")
```
